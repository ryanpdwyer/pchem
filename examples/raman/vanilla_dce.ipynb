# Read and plot one Raman CSV

Run from this notebook’s folder. This archived spectrum is a suspected DCE sample with an uncertain original label; replace it with the fresh class reference. No helper function or normalization is needed.

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

filename = "archived_dce_candidate.csv"
```

First inspect the header: integration time and scan averaging are recorded above the spectrum.

```python
with open(filename) as file:
    for i, line in enumerate(file):
        if i > 35:
            break
        print(i, line.rstrip())
```

For this file, 34 lines precede the column headings. For a new export, check the number rather than assuming it is unchanged.

```python
df = pd.read_csv(filename, skiprows=34)
df.head()
```

```python
plt.plot(df["Wavenumber"], df["Processed"])
plt.xlabel("Raman shift (cm⁻¹)")
plt.ylabel("Processed signal (counts)")
plt.show()
```

```python
plt.plot(df["Wavenumber"], df["Processed"], ".-", markersize=3)
plt.xlim(580, 820)
plt.xlabel("Raman shift (cm⁻¹)")
plt.ylabel("Processed signal (counts)")
plt.show()
```
